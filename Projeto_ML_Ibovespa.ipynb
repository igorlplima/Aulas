````python
# Criar uma nova versão do DataFrame para aplicar a correção histórica do Plano Real
df_corrigido = df_plot.copy()

# O Plano Real entrou em vigor em 01/07/1994.
# Todos os valores anteriores a esta data precisam ser divididos por 2.750 para serem convertidos para a escala do Real.
limite_plano_real = pd.to_datetime('1994-07-01')
fator_conversao = 2750.0

colunas_financeiras = ['Último', 'Abertura', 'Máxima', 'Mínima']

for col in colunas_financeiras:
    # Aplica a divisão apenas para as datas estritamente anteriores a julho de 1994
    df_corrigido.loc[df_corrigido['Data'] < limite_plano_real, col] = df_corrigido.loc[df_corrigido['Data'] < limite_plano_real, col] / fator_conversao

# Exibir os dados do período de transição (1994) para verificar se a inconsistência sumiu
display(df_corrigido[(df_corrigido['Data'] >= '1993-12-01') & (df_corrigido['Data'] <= '1994-08-01')])
````

````python
# Gerar novamente o gráfico de linha com a série histórica corrigida
plt.figure(figsize=(14, 7))
sns.set_theme(style="whitegrid")

sns.lineplot(data=df_corrigido, x='Data', y='Último', color='#2ca02c', linewidth=2)

plt.title('Histórico do Preço de Fechamento do Ibovespa (Corrigido pelo Plano Real)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Ano', fontsize=12)
plt.ylabel('Preço de Fechamento (Pontos Ajustados)', fontsize=12)

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
````

````python
import numpy as np

# Filtrar dados a partir do ano 2000 para melhor visualização
df_recente = df_corrigido[df_corrigido['Data'] >= '2000-01-01'].copy().reset_index(drop=True)

# Calcular Médias Móveis (Ex: 12 meses e 24 meses)
df_recente['MA12'] = df_recente['Último'].rolling(window=12).mean()
df_recente['MA24'] = df_recente['Último'].rolling(window=24).mean()

# Fazer uma regressão linear simples com os últimos 12 meses para projetar o próximo mês (Visualização de tendência)
ultimos_meses = df_recente.tail(12)
x = np.arange(len(ultimos_meses))
y = ultimos_meses['Último'].values
slope, intercept = np.polyfit(x, y, 1)

# Próximo ponto projetado
proximo_x = len(ultimos_meses)
proximo_y = slope * proximo_x + intercept
proxima_data = ultimos_meses['Data'].max() + pd.DateOffset(months=1)

# Exibir a projeção de curto prazo baseada na tendência recente
print(f"Último valor real registrado ({ultimos_meses['Data'].max().strftime('%m/%Y')}): {y[-1]:.3f} pontos")
print(f"Projeção linear simples para o próximo mês ({proxima_data.strftime('%m/%Y')}): {proximo_y:.3f} pontos (baseada na tendência de 12 meses)")
````

````python
# Gerar o gráfico interativo de análise de tendência
plt.figure(figsize=(15, 8))

# Plotar preço real
plt.plot(df_recente['Data'], df_recente['Último'], label='Preço de Fechamento Real', color='#2ca02c', alpha=0.5, linewidth=1.5)

# Plotar Médias Móveis
plt.plot(df_recente['Data'], df_recente['MA12'], label='Média Móvel 12 Meses (Tendência Curto/Médio Prazo)', color='#ff7f0e', linestyle='--', linewidth=2)
plt.plot(df_recente['Data'], df_recente['MA24'], label='Média Móvel 24 Meses (Tendência Longo Prazo)', color='#d62728', linestyle=':', linewidth=2)

# Plotar o ponto projetado para o mês seguinte
plt.scatter(proxima_data, proximo_y, color='blue', s=100, zorder=5, label=f'Projeção de Tendência ({proxima_data.strftime("%m/%Y")})')
plt.annotate(f'{proximo_y:.1f}', (proxima_data, proximo_y), textcoords="offset points", xytext=(0,10), ha='center', fontweight='bold', color='blue')

plt.title('Análise de Tendência e Projeção do Ibovespa (Desde 2000)', fontsize=16, fontweight='bold')
plt.xlabel('Ano', fontsize=12)
plt.ylabel('Pontuação', fontsize=12)
plt.legend(loc='upper left', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Valores reais (y) e previstos (y_pred) para o período de 12 meses analisado
y_real = y
y_pred = slope * x + intercept

# Cálculo das métricas de desempenho
mae = mean_absolute_error(y_real, y_pred)
mse = mean_squared_error(y_real, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_real, y_pred)

# Exibir os resultados formatados
print("=== Métricas de Avaliação do Modelo de Tendência Linear (Últimos 12 Meses) ===")
print(f"Erro Médio Absoluto (MAE): {mae:.3f} pontos")
print(f"Raiz do Erro Quadrático Médio (RMSE): {rmse:.3f} pontos")
print(f"Coeficiente de Determinação (R²): {r2:.4f}")
if r2 > 0.7:
    print("\nInterpretação: O modelo linear apresenta um ótimo ajuste aos dados recentes (R² forte).")
elif r2 > 0.4:
    print("\nInterpretação: O modelo linear captura moderadamente a tendência dos dados recentes.")
else:
    print("\nInterpretação: O ajuste linear é fraco devido à alta volatilidade recente do mercado.")
````

### Análise Complementar: O Impacto da Taxa Selic no Ibovespa
Para entender a dinâmica de longo prazo da nossa bolsa, vamos buscar os dados históricos da **Taxa Selic acumulada ao mês (% a.m.)** direto da API do Banco Central do Brasil (Série SGS 4390) e cruzá-los com o Ibovespa corrigido.

````python
import pandas as pd
import requests

# Código da série da Selic acumulada no mês no SGS/BCB: 4390
def obter_dados_sgs(codigo_serie, data_inicio='01/01/2000'):
    url = f"https://api.bcb.gov.br/dados/serie/bcdata.sgs.{codigo_serie}/dados?formato=json&dataInicial={data_inicio}"
    response = requests.get(url)
    if response.status_code == 200:
        df_selic = pd.DataFrame(response.json())
        df_selic['data'] = pd.to_datetime(df_selic['data'], format='%d/%m/%Y')
        df_selic['valor'] = pd.to_numeric(df_selic['valor'])
        return df_selic
    else:
        raise Exception("Erro ao acessar a API do Banco Central")

try:
    # Buscar dados da Selic a partir do ano 2000
    df_selic = obter_dados_sgs(4390, '01/01/2000')
    # Renomear colunas para o merge
    df_selic.rename(columns={'data': 'Data', 'valor': 'Selic_Mensal_%'}, inplace=True)

    # Como a Selic está em taxa mensal, vamos também anualizar de forma simples para fins de visualização intuitiva
    df_selic['Selic_Anualizada_%'] = ((1 + df_selic['Selic_Mensal_%']/100) ** 12 - 1) * 100

    display(df_selic.head())
except Exception as e:
    print("Não foi possível obter os dados da API:", e)
````

````python
# Fazer o merge dos dados do Ibovespa Recente com os dados da Selic
df_analise_macro = pd.merge(df_recente, df_selic, on='Data', how='inner')

# Criar gráfico de dois eixos (Y duplo) para comparar a evolução temporal
fig, ax1 = plt.subplots(figsize=(15, 7))

sns.set_theme(style="white")

# Primeiro eixo: Ibovespa
color = '#2ca02c'
ax1.set_xlabel('Ano', fontsize=12)
ax1.set_ylabel('Ibovespa (Pontos)', color=color, fontsize=12)
line1 = ax1.plot(df_analise_macro['Data'], df_analise_macro['Último'], color=color, linewidth=2, label='Ibovespa (Eixo Esq.)')
ax1.tick_params(axis='y', labelcolor=color)

# Segundo eixo: Selic Anualizada
ax2 = ax1.twinx()
color = '#d62728'
ax2.set_ylabel('Taxa Selic Anualizada (%)', color=color, fontsize=12)
line2 = ax2.plot(df_analise_macro['Data'], df_analise_macro['Selic_Anualizada_%'], color=color, linestyle='--', linewidth=1.8, label='Selic Anualizada (Eixo Dir.)')
ax2.tick_params(axis='y', labelcolor=color)

# Ajustar títulos e layout
plt.title('Comparativo Histórico: Ibovespa vs. Taxa Selic (Desde 2000)', fontsize=16, fontweight='bold', pad=15)
fig.tight_layout()

# Combinar as legendas de ambos os eixos
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left', fontsize=11)

plt.show()

# Calcular correlação de Pearson entre Ibovespa e Selic
correlacao = df_analise_macro['Último'].corr(df_analise_macro['Selic_Anualizada_%'])
print(f"\nCoeficiente de Correlação de Pearson entre o Ibovespa e a Selic: {correlacao:.4f}")
````

### Investigação e Tratamento de Valores Zerados e Ausentes
Investigando os dados históricos, identificamos a presença de valores iguais a zero nas colunas principais (`Último`, `Abertura`, etc.) correspondentes aos anos de 1989 e 1990. Vamos mapear esses registros e realizar o tratamento de remoção/limpeza.

````python
# Identificar onde o valor de fechamento é zero ou nulo
registros_zerados = df_corrigido[df_corrigido['Último'] == 0.0]
print(f"Quantidade de registros com valor igual a zero: {len(registros_zerados)}")
print("Período afetado pelos valores zerados:")
display(registros_zerados[['Data', 'Último']].head(10))

# Filtrar a base mantendo apenas registros onde a pontuação seja estritamente maior que zero
df_limpo = df_corrigido[df_corrigido['Último'] > 0.0].copy().reset_index(drop=True)
print(f"\nDimensões após remover valores nulos/zerados: {df_limpo.shape}")
print(f"Nova data inicial da série histórica consistente: {df_limpo['Data'].min().strftime('%d/%m/%Y')}")
````

### Modelagem de Série Temporal, Validação Temporal (TimeSeriesSplit) e Otimização de Hiperparâmetros
Nesta seção, vamos estruturar a base de dados para um problema de aprendizado supervisionado (criando lags/atrasos temporais) e utilizar validação cruzada temporal adequada para otimizar os parâmetros de um estimador de Floresta Aleatória (`RandomForestRegressor`).

````python
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.ensemble import RandomForestRegressor

# 1. Criação de Features Temporais (Lags de 1, 2 e 3 meses)
df_features = df_limpo[['Data', 'Último']].copy()
df_features['Lag_1'] = df_features['Último'].shift(1)
df_features['Lag_2'] = df_features['Último'].shift(2)
df_features['Lag_3'] = df_features['Último'].shift(3)

# Remover as linhas com NaNs decorrentes dos deslocamentos (lags)
df_features.dropna(inplace=True)
df_features.reset_index(drop=True, inplace=True)

# Definir variáveis preditoras (X) e alvo (y)
X = df_features[['Lag_1', 'Lag_2', 'Lag_3']]
y = df_features['Último']

# 2. Separação Estrita do Conjunto de Teste Final (Últimos 12 meses - Out-of-Sample)
# Este conjunto NÃO será visto na validação cruzada nem na otimização de hiperparâmetros
X_train_cv, X_test = X.iloc[:-12], X.iloc[-12:]
y_train_cv, y_test = y.iloc[:-12], y.iloc[-12:]

print(f"Tamanho do treino/validação histórica (CV): {X_train_cv.shape[0]} meses")
print(f"Tamanho do teste final blindado: {X_test.shape[0]} meses (Período: {df_features['Data'].iloc[-12:].min().strftime('%m/%Y')} a {df_features['Data'].iloc[-12:].max().strftime('%m/%Y')})")
````

````python
# 3. Configuração do TimeSeriesSplit e Grade de Hiperparâmetros
tscv = TimeSeriesSplit(n_splits=5)

# Definir estimador base
rf = RandomForestRegressor(random_state=42)

# Grade de parâmetros para testar sem tocar no teste final
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [3, 5, 8, None],
    'min_samples_split': [2, 5, 10]
}

# Executar GridSearchCV com validação temporal cruzada
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv=tscv, scoring='neg_mean_absolute_error', n_jobs=-1)
grid_search.fit(X_train_cv, y_train_cv)

# Melhores parâmetros encontrados de forma justa
print("=== Resultados da Otimização de Hiperparâmetros (Validação Temporal) ===")
print("Melhores Hiperparâmetros:", grid_search.best_params_)
print(f"Melhor MAE obtido em validação: {abs(grid_search.best_score_):.3f} pontos")
````

### Teste Final (Out-of-Sample) e Avaliação de Desempenho
Agora avaliaremos o modelo otimizado no conjunto de teste que mantivemos blindado, calculando as métricas finais (MAE, RMSE e R²) e exibindo um comparativo visual dos últimos 12 meses de dados reais contra as previsões.

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Recuperar o melhor modelo otimizado pelo GridSearchCV
melhor_modelo = grid_search.best_estimator_

# Fazer as predições no conjunto de teste final (blindado)
previsoes_teste = melhor_modelo.predict(X_test)

# Calcular as métricas finais de teste
mae_teste = mean_absolute_error(y_test, previsoes_teste)
rmse_teste = np.sqrt(mean_squared_error(y_test, previsoes_teste))
r2_teste = r2_score(y_test, previsoes_teste)

print("=== Métricas de Avaliação no Teste Final (Out-of-Sample) ===")
print(f"MAE Final: {mae_teste:.3f} pontos")
print(f"RMSE Final: {rmse_teste:.3f} pontos")
print(f"R² de Teste Final: {r2_teste:.4f}")
````

````python
# Criar DataFrame temporário para plotar o teste final
datas_teste = df_features['Data'].iloc[-12:]
df_comparacao = pd.DataFrame({
    'Data': datas_teste,
    'Real': y_test,
    'Previsto': previsoes_teste
}).reset_index(drop=True)

# Plotar gráfico comparando o Real vs Previsto
plt.figure(figsize=(12, 6))
plt.plot(df_comparacao['Data'], df_comparacao['Real'], label='Real', color='black', marker='o', linewidth=2)
plt.plot(df_comparacao['Data'], df_comparacao['Previsto'], label='Previsão Random Forest', color='darkorange', marker='s', linestyle='--', linewidth=2)

plt.title('Avaliação Final (Out-of-Sample): Real vs. Previsão do Ibovespa', fontsize=14, fontweight='bold')
plt.xlabel('Data', fontsize=12)
plt.ylabel('Pontuação', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)

# Inserir as métricas em uma caixa de texto
texto_final = f"Métricas de Teste:\nMAE: {mae_teste:.2f} pts\nRMSE: {rmse_teste:.2f} pts\nR²: {r2_teste:.4f}"
plt.gca().text(0.05, 0.95, texto_final, transform=plt.gca().transAxes, fontsize=10,
            verticalalignment='top', bbox=dict(boxstyle='round,pad=0.5', facecolor='white', alpha=0.8, edgecolor='orange'))

plt.tight_layout()
plt.show()
````
